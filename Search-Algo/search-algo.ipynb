## Linear Search

```python
# Linear Search 

# Brute Force

def linear_search(arr: list, item:int):
    for i in range(len(arr)):
        if arr[i] == item:
            return i
    
    return -1
```

```python
arrary = [87,9,45,74,14,52,6,5,48,45,2,33,4,8,]

linear_search(arrary,77)
```

## Binary Search

```python
# Sorted Array

def binary_search(arr, low, high, item):
    if low <= high:
        mid = low + high //2
        if arr[mid] == item:
            return mid
        elif arr[mid] > item:
            return binary_search(arr, low, mid-1, item)
        else:
            return binary_search(arr, mid-1,high,item)
    else:
        return -1
```

```python
arr =[5,7,17,19,22,29,38,49,56,68,79,85,97]

binary_search(arr,0,len(arr)-1,38)
```

```python
##  Check whether the given array is sorted or Not.

def is_sorted(arr):
    for i in range(len(arr)-1):
        if arr[i] > arr[i+1]:
            return False
    return True

arr = [1,2,6,8,19,20,25,84,94]

is_sorted(arr)
```
