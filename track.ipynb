```python
def strStr(haystack: str, needle: str) -> int:
    nlen = len(needle)
    j = 0
    print(f'len of needle: {nlen}')
    for i in range(len(haystack)):
        
        print(f'i {i}\nj {j}')
        if j <= nlen:
            print(f' str = {haystack[i]}  &  needle = {needle[j]} ')
            if haystack[i] == needle[j]:
                j += 1
                print(f' updated :- j { j} ')
                if j == nlen:
                    return i - nlen + 1
            else:
                j = 0  
                print('J set to 0')  
    return -1
```
