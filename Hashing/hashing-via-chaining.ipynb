```python
class Node:
    def __init__(self,key,value):
        self.key = key
        self.value = value
        self.next = None
```

```python
class LinkedList:
    
    def __init__(self):
        
        # Empty Linked List
        self.head = None
        # No. of nodes in the LL
        self.n = 0
        
    def __len__(self):
        return self.n
        
        
    def __str__(self):
        curr = self.head
        result =''
        while curr != None:
            result = f'{result} {str(curr.data)} ->'
            curr = curr.next
        return result[:-2]
    
    
    def append(self,key,value):
         # New node
        new_node = Node(key,value)
        if self.head == None:
            # Empty
            self.head = new_node
            self.n += 1
            return 
    
        # Create connection
        curr = self.head
        
        while curr.next != None:
            curr = curr.next
        
        # you are at last node
        curr.next = new_node
       
        # Increament n
        self.n += 1
        
    def clear(self):
        self.head = None
        self.n = 0
     
    def delete_head(self):
            if self.head == None:
                return 'Empty Linked List'
            self.head = self.head.next
            self.n -= 1
              
    
    def remove(self,key):
        # If linked list is empty
        if self.head == None:
            return 'Empty linked List'
        
        # Removing head if the value matchs
        if self.head.key == key:
            return self.delete_head()
        
        curr = self.head
        while curr.next != None:
            if curr.next.key == key:
                break
            curr = curr.next
        
        # case 1 (item found):-
        if curr.next != None:
            curr.next = curr.next.next
            self.n -= 1
        else:
            # case 2 (item not found):-
            return 'Item not found'
    
    def seach(self,key):
        curr = self.head
        pos = 0
        
        while curr != None:
            if curr.key == key:
                return pos
            
            curr = curr.next
            pos += 1
        
        return -1
    
    def __getitem__(self,index):
        curr = self.head
        pos = 0
        
        while curr != None:
            if pos == index:
                return curr.data
            curr = curr.next
            pos +=  1
            
        return 'IndexError'
    
    def traverse(self):
        temp = self.head
        while temp != None:
            print(temp.key,'-->', temp.value ," ",end=' --> ')
            temp = temp.next
```
